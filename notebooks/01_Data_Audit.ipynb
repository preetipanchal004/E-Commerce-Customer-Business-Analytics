```python
import pandas as pd
import numpy as np

dataset_path = r"C:\Users\user\Downloads\Indian E-Commerce Sales & Customer Analytics"

if pd.io.common.os.path.exists(dataset_path):
    print("Dataset folder connected successfully!")
else:
    print("Dataset folder not found. Check the folder path.")
```

```python
# List of required CSV files

required_files = [
    "customers.csv",
    "products.csv",
    "orders.csv",
    "order_items.csv",
    "payments.csv",
    "returns.csv",
    "shipments.csv",
    "customer_reviews.csv",
    "marketing_campaigns.csv"
]

# Check file availability

available_files = pd.io.common.os.listdir(dataset_path)

for file in required_files:
    if file in available_files:
        print(f"{file} - Available")
    else:
        print(f"{file} - Missing")
```

```python
customers = pd.read_csv(f"{dataset_path}\\customers.csv")
print("CUSTOMERS DATASET AUDIT","-" * 50)

print("Dataset Shape:", customers.shape)

print("\nColumn Names:",list(customers.columns))

print("\nData Types:",customers.dtypes)

print("\nFirst 5 Records:",customers.head())
```

```python
print("MISSING VALUES","-" * 50)

print(customers.isnull().sum())

print("\nTotal Missing Values:", customers.isnull().sum().sum())
print("\nDUPLICATE RECORDS","-" * 50)

print("Duplicate Rows:", customers.duplicated().sum())

print("Duplicate Customer IDs:",customers["customer_id"].duplicated().sum())
```

```python
# Customers: Unique Values Audit

columns_to_check = [
    "gender",
    "age_group",
    "customer_segment",
    "preferred_device",
    "preferred_payment_method",
    "acquisition_channel",
    "customer_status",
    "loyalty_tier"
]

print("CUSTOMER CATEGORY VALIDATION","-" * 50)

for column in columns_to_check:
    print(f"\n{column}")
    print(customers[column].value_counts(dropna=False))

print("\nCUSTOMER ID VALIDATION")
print("-" * 50)

print("Unique Customer IDs:", customers["customer_id"].nunique())
print("Total Customer Records:", len(customers))
print("First Customer ID:", customers["customer_id"].min())
print("Last Customer ID:", customers["customer_id"].max())
```

```python
# Age and Age Group Validation

print("AGE VALIDATION")
print("-" * 50)

print("Minimum Age:", customers["age"].min())
print("Maximum Age:", customers["age"].max())
print("Average Age:", customers["age"].mean())
print("Missing Ages:", customers["age"].isnull().sum())

print("\nAGE GROUP DISTRIBUTION")
print("-" * 50)

print(customers.groupby("age_group")["age"].agg(
    ["count", "min", "max"]
))

print("\nINVALID AGE RECORDS")
print("-" * 50)

invalid_age = customers[
    (customers["age"] < 18) | (customers["age"] > 100)
]

print("Invalid Age Records:", len(invalid_age))
```

```python
# Customer Date Columns Audit

date_columns = [
    "customer_signup_date",
    "last_order_date"
]

print("CUSTOMER DATE VALIDATION")
print("-" * 50)

for column in date_columns:
    dates = pd.to_datetime(customers[column], errors="coerce")

    print(f"\n{column}")
    print("Minimum Date:", dates.min())
    print("Maximum Date:", dates.max())
    print("Invalid Dates:", dates.isnull().sum() - customers[column].isnull().sum())
    print("Missing Dates:", customers[column].isnull().sum())
```

```python
# Customer Date and Status Audit

customers["customer_signup_date"] = pd.to_datetime(
    customers["customer_signup_date"], errors="coerce"
)

customers["last_order_date"] = pd.to_datetime(
    customers["last_order_date"], errors="coerce"
)

print("CUSTOMER DATE & STATUS AUDIT")
print("-" * 50)

print("Signup Date Range:",
      customers["customer_signup_date"].min(),
      "to",
      customers["customer_signup_date"].max())

print("\nLast Order Date Range:",
      customers["last_order_date"].min(),
      "to",
      customers["last_order_date"].max())

print("\nCustomers with 0 Orders:",
      (customers["total_orders"] == 0).sum())

print("Customers with Missing Last Order Date:",
      customers["last_order_date"].isnull().sum())

print("\nCustomer Status vs Orders")
print(pd.crosstab(
    customers["customer_status"],
    customers["total_orders"].eq(0)
))
```

```python
# Products Dataset Audit

products = pd.read_csv(f"{dataset_path}\\products.csv")

print("PRODUCTS DATASET AUDIT")
print("-" * 50)

print("Dataset Shape:", products.shape)

print("\nColumn Names:")
print(list(products.columns))

print("\nData Types:")
print(products.dtypes)

print("\nFirst 5 Records:")
display(products.head())
```

```python
# Products Missing Values, Duplicates & Category Audit

print("MISSING VALUES")
print("-" * 40)
print(products.isnull().sum())
print("Total Missing:", products.isnull().sum().sum())

print("\nDUPLICATE AUDIT")
print("-" * 40)
print("Duplicate Rows:", products.duplicated().sum())
print("Duplicate Product IDs:", products["product_id"].duplicated().sum())

print("\nCATEGORY DISTRIBUTION")
print("-" * 40)

for col in ["category", "subcategory", "brand", "product_type"]:
    print(f"\n{col}:")
    print(products[col].value_counts(dropna=False))

print("\nPRICE VALIDATION")
print("-" * 40)
print(products[["price", "cost_price", "stock_quantity"]].describe())
```

```python
# Product Business Logic Validation

print("PRODUCT VALIDATION")
print("-" * 45)

print("Products with Price <= 0:",
      (products["price"] <= 0).sum())

print("Products with Cost Price <= 0:",
      (products["cost_price"] <= 0).sum())

print("Products where Cost Price >= Selling Price:",
      (products["cost_price"] >= products["price"]).sum())

print("Products with Negative Stock:",
      (products["stock_quantity"] < 0).sum())

print("Invalid Rating Values:",
      ((products["rating_average"] < 0) |
       (products["rating_average"] > 5)).sum())

print("\nProduct Launch Date Range:")
print(
    pd.to_datetime(products["product_launch_date"], errors="coerce").min(),
    "to",
    pd.to_datetime(products["product_launch_date"], errors="coerce").max()
)
```

```python
# Orders Dataset Audit

orders = pd.read_csv(f"{dataset_path}\\orders.csv")

print("ORDERS DATASET AUDIT")
print("-" * 45)

print("Dataset Shape:", orders.shape)

print("\nColumn Names:")
print(list(orders.columns))

print("\nData Types:")
print(orders.dtypes)

print("\nFirst 5 Records:")
display(orders.head())
```

```python
# Orders: Missing, Duplicate, Category & Amount Audit

print("MISSING VALUES")
print("-" * 40)
print(orders.isnull().sum())
print("Total Missing:", orders.isnull().sum().sum())

print("\nDUPLICATES")
print("-" * 40)
print("Duplicate Rows:", orders.duplicated().sum())
print("Duplicate Order IDs:", orders["order_id"].duplicated().sum())

print("\nORDER STATUS")
print(orders["order_status"].value_counts(dropna=False))

print("\nSHIPPING METHOD")
print(orders["shipping_method"].value_counts(dropna=False))

print("\nMARKETING CHANNEL")
print(orders["marketing_channel"].value_counts(dropna=False))

print("\nAMOUNT VALIDATION")
print("Subtotal <= 0:", (orders["subtotal"] <= 0).sum())
print("Negative Shipping Fee:", (orders["shipping_fee"] < 0).sum())
print("Negative Tax:", (orders["tax_amount"] < 0).sum())
print("Negative Final Amount:", (orders["final_amount"] < 0).sum())

expected_amount = (
    orders["subtotal"] +
    orders["shipping_fee"] +
    orders["tax_amount"]
)

print("Final Amount Mismatch:",
      (~(abs(orders["final_amount"] - expected_amount) < 0.01)).sum())
```

```python
# Order Items Dataset Audit

order_items = pd.read_csv(f"{dataset_path}\\order_items.csv")

print("ORDER ITEMS DATASET AUDIT")
print("-" * 45)

print("Dataset Shape:", order_items.shape)

print("\nColumn Names:")
print(list(order_items.columns))

print("\nData Types:")
print(order_items.dtypes)

print("\nFirst 5 Records:")
display(order_items.head())
```

```python
# Order Items: Complete Basic Audit

print("MISSING VALUES")
print("-" * 40)
print(order_items.isnull().sum())
print("Total Missing:", order_items.isnull().sum().sum())

print("\nDUPLICATES")
print("-" * 40)
print("Duplicate Rows:", order_items.duplicated().sum())
print("Duplicate Item IDs:",
      order_items["order_item_id"].duplicated().sum())

print("\nBUSINESS VALIDATION")
print("-" * 40)

print("Quantity <= 0:", (order_items["quantity"] <= 0).sum())
print("Unit Price <= 0:", (order_items["unit_price"] <= 0).sum())
print("Negative Item Revenue:", (order_items["item_revenue"] < 0).sum())
print("Negative Item Cost:", (order_items["item_cost"] < 0).sum())

expected_revenue = (
    order_items["quantity"] *
    order_items["unit_price"] *
    (1 - order_items["discount_percentage"] / 100)
)

print("Revenue Mismatch:",
      (abs(order_items["item_revenue"] - expected_revenue) > 0.02).sum())

expected_profit = order_items["item_revenue"] - order_items["item_cost"]

print("Profit Mismatch:",
      (abs(order_items["profit"] - expected_profit) > 0.02).sum())

print("\nDiscount Range:")
print(order_items["discount_percentage"].min(),
      "to",
      order_items["discount_percentage"].max())
```

```python
# Shipments Complete Audit

print("MISSING VALUES")
print("-" * 40)
print(shipments.isnull().sum())
print("Total Missing:", shipments.isnull().sum().sum())

print("\nDUPLICATES")
print("-" * 40)
print("Duplicate Rows:", shipments.duplicated().sum())
print("Duplicate Shipment IDs:", shipments["shipment_id"].duplicated().sum())
print("Duplicate Order IDs:", shipments["order_id"].duplicated().sum())

print("\nDELIVERY STATUS")
print(shipments["delivery_status"].value_counts(dropna=False))

print("\nDELAYED FLAG")
print(shipments["delayed_flag"].value_counts(dropna=False))

print("\nBUSINESS VALIDATION")
print("-" * 40)
print("Invalid Delayed Flags:",
      (~shipments["delayed_flag"].isin([0, 1])).sum())

print("Negative Delivery Days:",
      (shipments["delivery_days"] < 0).sum())

print("\nDATE VALIDATION")

for col in ["dispatch_date", "expected_delivery_date", "actual_delivery_date"]:
    dates = pd.to_datetime(shipments[col], errors="coerce")
    print(col, "Invalid Dates:",
          (dates.isnull() & shipments[col].notnull()).sum())
```

```python
# Returns Dataset - Basic Audit

returns = pd.read_csv(dataset_path + r"\returns.csv")

print("DATASET SHAPE:", returns.shape)

print("\nCOLUMNS:")
print(returns.columns.tolist())

print("\nDATA TYPES:")
print(returns.dtypes)

print("\nMISSING VALUES:")
print(returns.isnull().sum())
print("Total Missing:", returns.isnull().sum().sum())

print("\nDUPLICATES:")
print("Duplicate Rows:", returns.duplicated().sum())
print("Duplicate Return IDs:", returns["return_id"].duplicated().sum())

print("\nRETURN STATUS:")
print(returns["return_status"].value_counts(dropna=False))

print("\nRETURN REASONS:")
print(returns["return_reason"].value_counts(dropna=False))

print("\nFIRST 5 RECORDS:")
display(returns.head())
```

```python
# Returns Business Validation

print("REFUND VALIDATION")
print("-" * 40)
print("Negative Refunds:", (returns["refund_amount"] < 0).sum())
print("Zero Refunds:", (returns["refund_amount"] == 0).sum())

print("\nREFUND AMOUNT BY STATUS")
print(returns.groupby("return_status")["refund_amount"].agg(["count", "min", "max", "mean"]).round(2))

print("\nRETURN DATE VALIDATION")
dates = pd.to_datetime(returns["return_date"], errors="coerce")
print("Invalid Dates:", (dates.isnull() & returns["return_date"].notnull()).sum())
print("Date Range:", dates.min(), "to", dates.max())
```

```python
# Customer Reviews Basic Audit

reviews = pd.read_csv(dataset_path + r"\customer_reviews.csv")

print("DATASET SHAPE:", reviews.shape)
print("\nCOLUMNS:")
print(reviews.columns.tolist())

print("\nDATA TYPES:")
print(reviews.dtypes)

print("\nMISSING VALUES:")
print(reviews.isnull().sum())
print("Total Missing:", reviews.isnull().sum().sum())

print("\nDUPLICATES:")
print("Duplicate Rows:", reviews.duplicated().sum())
print("Duplicate Review IDs:", reviews["review_id"].duplicated().sum())

print("\nFIRST 5 RECORDS:")
display(reviews.head())
```

```python
# Customer Reviews Business Validation

print("RATING DISTRIBUTION")
print(reviews["rating"].value_counts().sort_index())

print("\nINVALID RATINGS:", (~reviews["rating"].between(1, 5)).sum())

print("\nSENTIMENT DISTRIBUTION")
print(reviews["review_sentiment"].value_counts(dropna=False))

print("\nRATING vs SENTIMENT")
print(pd.crosstab(reviews["rating"], reviews["review_sentiment"], dropna=False))

print("\nVERIFIED PURCHASE")
print(reviews["verified_purchase"].value_counts())

print("\nDATE VALIDATION")
dates = pd.to_datetime(reviews["review_date"], errors="coerce")
print("Invalid Dates:", (dates.isnull() & reviews["review_date"].notnull()).sum())
print("Date Range:", dates.min(), "to", dates.max())
```

```python
# Marketing Campaigns Basic Audit

campaigns = pd.read_csv(dataset_path + r"\marketing_campaigns.csv")

print("DATASET SHAPE:", campaigns.shape)

print("\nCOLUMNS:")
print(campaigns.columns.tolist())

print("\nDATA TYPES:")
print(campaigns.dtypes)

print("\nMISSING VALUES:")
print(campaigns.isnull().sum())
print("Total Missing:", campaigns.isnull().sum().sum())

print("\nDUPLICATES:")
print("Duplicate Rows:", campaigns.duplicated().sum())
print("Duplicate Campaign IDs:", campaigns["campaign_id"].duplicated().sum())

print("\nFIRST 5 RECORDS:")
display(campaigns.head())
```

```python
# Marketing Campaign Business Validation

print("DISCOUNT VALIDATION")
print("Invalid Discount (below 0 or above 100):",
      (~campaigns["discount_percentage"].between(0, 100)).sum())

print("\nCAMPAIGN METRICS VALIDATION")
print("Negative Conversions:", (campaigns["conversions"] < 0).sum())
print("Negative Impressions:", (campaigns["impressions"] < 0).sum())
print("Negative Clicks:", (campaigns["clicks"] < 0).sum())
print("Clicks Greater Than Impressions:",
      (campaigns["clicks"] > campaigns["impressions"]).sum())
print("Negative Revenue:", (campaigns["revenue_generated"] < 0).sum())
print("Negative Campaign Cost:", (campaigns["campaign_cost"] < 0).sum())

print("\nROI VALIDATION")
expected_roi = (
    (campaigns["revenue_generated"] - campaigns["campaign_cost"])
    / campaigns["campaign_cost"].replace(0, np.nan)
).round(3)

print("ROI Mismatches:", (campaigns["roi"] - expected_roi).abs().gt(0.001).sum())

print("\nDATE VALIDATION")
for col in ["campaign_start_date", "campaign_end_date"]:
    dates = pd.to_datetime(campaigns[col], errors="coerce")
    print(col, "Invalid Dates:",
          (dates.isnull() & campaigns[col].notnull()).sum())

start = pd.to_datetime(campaigns["campaign_start_date"])
end = pd.to_datetime(campaigns["campaign_end_date"])

print("End Date Before Start Date:", (end < start).sum())
print("Campaign Date Range:", start.min(), "to", end.max())
```
