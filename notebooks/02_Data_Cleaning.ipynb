```python
# E-Commerce Data Cleaning - Setup

import pandas as pd
import numpy as np

dataset_path = r"C:\Users\user\Downloads\Indian E-Commerce Sales & Customer Analytics"

output_path = r"C:\Users\user\Documents\E-Commerce-Customer-Business-Analytics\data\processed"

print("Data Cleaning Environment Ready!")
print("Raw Dataset Folder:", dataset_path)
print("Processed Dataset Folder:", output_path)
```

```python
# Customers Data Cleaning

customers = pd.read_csv(dataset_path + r"\customers.csv")

# Convert date columns
customers["customer_signup_date"] = pd.to_datetime(customers["customer_signup_date"])
customers["last_order_date"] = pd.to_datetime(customers["last_order_date"])

# Fill missing age with median
customers["age"] = customers["age"].fillna(customers["age"].median())

# Fill missing device
customers["preferred_device"] = customers["preferred_device"].fillna("Unknown")

# Save cleaned file
customers.to_csv(output_path + r"\customers_cleaned.csv", index=False)

print("Customers Data Cleaning Completed!")
print("Final Shape:", customers.shape)
print("Remaining Missing Values:", customers.isnull().sum().sum())
```

```python
# Products Data Cleaning

products = pd.read_csv(dataset_path + r"\products.csv")

products["product_launch_date"] = pd.to_datetime(products["product_launch_date"])

products["rating_average"] = products["rating_average"].fillna(
    products["rating_average"].median()
)

products.to_csv(output_path + r"\products_cleaned.csv", index=False)

print("Products Data Cleaning Completed!")
print("Final Shape:", products.shape)
print("Remaining Missing Values:", products.isnull().sum().sum())
```

```python
# Orders Data Cleaning

orders = pd.read_csv(dataset_path + r"\orders.csv")

orders["order_date"] = pd.to_datetime(orders["order_date"])

orders.to_csv(output_path + r"\orders_cleaned.csv", index=False)

print("Orders Data Cleaning Completed!")
print("Final Shape:", orders.shape)
print("Remaining Missing Values:", orders.isnull().sum().sum())
```

```python
# Order Items Data Cleaning

order_items = pd.read_csv(dataset_path + r"\order_items.csv")

order_items.to_csv(output_path + r"\order_items_cleaned.csv", index=False)

print("Order Items Data Cleaning Completed!")
print("Final Shape:", order_items.shape)
print("Remaining Missing Values:", order_items.isnull().sum().sum())
```

```python
# Payments Data Cleaning

payments = pd.read_csv(dataset_path + r"\payments.csv")

payments["payment_date"] = pd.to_datetime(payments["payment_date"])

payments.to_csv(output_path + r"\payments_cleaned.csv", index=False)

print("Payments Data Cleaning Completed!")
print("Final Shape:", payments.shape)
print("Remaining Missing Values:", payments.isnull().sum().sum())
```

```python
# Shipments Data Cleaning

shipments = pd.read_csv(dataset_path + r"\shipments.csv")

date_columns = ["dispatch_date", "expected_delivery_date", "actual_delivery_date"]

for col in date_columns:
    shipments[col] = pd.to_datetime(shipments[col])

shipments.to_csv(output_path + r"\shipments_cleaned.csv", index=False)

print("Shipments Data Cleaning Completed!")
print("Final Shape:", shipments.shape)
print("Remaining Missing Values:", shipments.isnull().sum().sum())
```

```python
# Returns Data Cleaning

returns = pd.read_csv(dataset_path + r"\returns.csv")

returns["return_date"] = pd.to_datetime(returns["return_date"])

returns.to_csv(output_path + r"\returns_cleaned.csv", index=False)

print("Returns Data Cleaning Completed!")
print("Final Shape:", returns.shape)
print("Remaining Missing Values:", returns.isnull().sum().sum())
```

```python
# Customer Reviews Data Cleaning

reviews = pd.read_csv(dataset_path + r"\customer_reviews.csv")

reviews["review_date"] = pd.to_datetime(reviews["review_date"])

sentiment_map = {
    1: "Negative",
    2: "Negative",
    3: "Neutral",
    4: "Positive",
    5: "Positive"
}

reviews["review_sentiment"] = reviews["review_sentiment"].fillna(
    reviews["rating"].map(sentiment_map)
)

reviews.to_csv(output_path + r"\customer_reviews_cleaned.csv", index=False)

print("Customer Reviews Data Cleaning Completed!")
print("Final Shape:", reviews.shape)
print("Remaining Missing Values:", reviews.isnull().sum().sum())
```

```python
# Marketing Campaigns Data Cleaning

campaigns = pd.read_csv(dataset_path + r"\marketing_campaigns.csv")

campaigns["campaign_start_date"] = pd.to_datetime(campaigns["campaign_start_date"])
campaigns["campaign_end_date"] = pd.to_datetime(campaigns["campaign_end_date"])

campaigns.to_csv(output_path + r"\marketing_campaigns_cleaned.csv", index=False)

print("Marketing Campaigns Data Cleaning Completed!")
print("Final Shape:", campaigns.shape)
print("Remaining Missing Values:", campaigns.isnull().sum().sum())
```

```python
# Final Cleaned Files Verification

files = [
    "customers_cleaned.csv",
    "products_cleaned.csv",
    "orders_cleaned.csv",
    "order_items_cleaned.csv",
    "payments_cleaned.csv",
    "shipments_cleaned.csv",
    "returns_cleaned.csv",
    "customer_reviews_cleaned.csv",
    "marketing_campaigns_cleaned.csv"
]

for file in files:
    df = pd.read_csv(output_path + "\\" + file)
    print(file, "->", df.shape)
```
